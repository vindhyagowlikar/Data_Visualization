```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("weather_data.csv")

print("WEEK 4: DATA VISUALIZATION")

# Line Plot
plt.figure(figsize=(10, 5))

plt.plot(
    df["Date"],
    df["Temperature"],
    marker="o"
)

plt.title("Temperature by Date")
plt.xlabel("Date")
plt.ylabel("Temperature")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("temperature_lineplot.png")

plt.show()


# Histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df["Temperature"],
    bins=6
)

plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig("temperature_histogram.png")

plt.show()


# Scatter Plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Humidity",
    y="Temperature"
)

plt.title("Humidity vs Temperature")

plt.tight_layout()

plt.savefig("humidity_temperature.png")

plt.show()


# Rainfall vs Sales
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rainfall",
    y="Sales"
)

plt.title("Rainfall vs Sales")

plt.tight_layout()

plt.savefig("rainfall_sales.png")

plt.show()


# Correlation Heatmap
plt.figure(figsize=(7, 5))

correlation = df[
    ["Temperature",
     "Humidity",
     "Rainfall",
     "Sales"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("correlation_heatmap.png")

plt.show()

print("All visualizations created successfully.")
```
